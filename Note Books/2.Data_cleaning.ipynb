```python
df1 = pd.read_csv("../data/accounts.csv")
print(df1.isnull().sum())
print("Duplicate count:",df1.duplicated().sum())
```

```python
# Select rows where any column is missing
display(df1[df1.isnull().any(axis=1)])
```
